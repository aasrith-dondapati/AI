```python
from nltk.util import bigrams,trigrams,ngrams
```

```python
string = 'hello the best and most beautifull thing in the world cannot be seen or even touched,they must be felt with heart'
quotes_tokens = nltk.word_tokenize(string)
quotes_tokens
```

```python
string
```

```python
quotes_tokens
```

```python
len(quotes_tokens)
```

```python
quotes_bigrams = list(nltk.bigrams(quotes_tokens))
quotes_bigrams
```

```python
quotes_trigrams = list(nltk.trigrams(quotes_tokens))
quotes_trigrams
```

```python
quotes_ngrams = list(nltk.ngrams(quotes_tokens, 4))
quotes_ngrams
```

```python
len(quotes_ngrams)
```

```python
from nltk.stem import PorterStemmer
pst = PorterStemmer()
```

```python
pst.stem('affection')
```

```python
pst.stem('playing')
```

```python
pst.stem('maximum')
```

```python
word_to_stem = ['give','giving','given','gived','thinking','loving','maximum']
for word in word_to_stem:
    print(word + ':'+pst.stem(word))
```

```python
from nltk.stem import LancasterStemmer
lst = LancasterStemmer()
word_to_stem = ['give','giving','given','gived','thinking','loving','maximum']
for word in word_to_stem:
    print(word + ':'+lst.stem(word))
```

```python
from nltk.stem import SnowballStemmer
sns = SnowballStemmer('english')
word_to_stem = ['give','giving','given','gived','thinking','loving','maximum']
for word in word_to_stem:
    print(word + ':'+sns.stem(word))
```
